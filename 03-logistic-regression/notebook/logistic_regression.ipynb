```python
def sigmoid(z):
    return 1/(1+np.exp(-z))
```

```python
def compute_fwb(X, w, b):
    z = np.dot(X, w) + b
    return sigmoid(z)
```

```python
def compute_cost(X, y, w, b, fwb_Function=compute_fwb):
    r, c = X.shape
    cost = 0
    for i in range(r):
        fwb_i = fwb_Function(X[i], w, b)
        cost += (-y[i] * np.log(fwb_i)) - ((1-y[i]) * np.log(1-fwb_i))

    return cost/r
```
