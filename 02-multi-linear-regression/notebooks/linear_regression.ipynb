```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import math
```

```python
from sklearn.datasets import fetch_california_housing

housing = fetch_california_housing(as_frame=True)
```

```python
data = housing.frame
```

```python
data.head()
```

```python
data.columns
```

```python
X = data.drop("MedHouseVal", axis=1)
y = data.MedHouseVal
```

```python
plt.scatter(X["MedInc"], y, marker="X", c="r")
plt.xlabel("Medien Income")
plt.ylabel("Median House Value")
plt.title("Medien Income VS Median House Value")
plt.show()
```

```python
plt.scatter(X["HouseAge"], y, marker="X", c="r")
plt.xlabel("House Age")
plt.ylabel("Median House Value")
plt.title("House Age VS Median House Value")
plt.show()
```

```python
plt.scatter(X["AveBedrms"], y, marker="X", c="r")
plt.xlabel("Average Number of Bedrooms")
plt.ylabel("Median House Value")
plt.title("Average Number of Bedrooms VS Median House Value")
plt.show()
```

```python
plt.scatter(X["Population"], y, marker="X", c="r")
plt.xlabel("Population")
plt.ylabel("Median House Value")
plt.title("Population VS Median House Value")
plt.show()
```

```python
plt.scatter(X["AveOccup"], y, marker="X", c="r")
plt.xlabel("Average Occupant")
plt.ylabel("Median House Value")
plt.title("Average Occupant VS Median House Value")
plt.show()
```

```python
plt.scatter(X["Latitude"], y, marker="X", c="r")
plt.xlabel("Latitude")
plt.ylabel("Median House Value")
plt.title("Latitude VS Median House Value")
plt.show()
```
